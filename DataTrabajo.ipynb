```python
# PASO 1 — Entorno: warnings, plt, np, pd, sns, display. whitegrid. versión de pandas.
import warnings
warnings.filterwarnings("ignore")
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from IPython.display import display
sns.set_theme(style="whitegrid", context="notebook")
pd.__version__
```

```python
# PASO 2 — Cargar csv y limpieza
raw = pd.read_csv("/content/cantidad_de_nacimientos_edad_madre_por_anio_de_nacimiento_2012_2024.csv")
df = raw.dropna().copy()
df = df.rename(columns ={
    "provincia_nombre":"provincia"
})
df.info()
df["grupo_etario_madre"].unique()
```

```python
fig, axes = plt.subplots(1, figsize=(12, 4))
sns.histplot(data=df, x='anio', weights='cantidad', bins=13, ax=axes, color="blue")
axes.ticklabel_format(style='plain', axis='y')
axes.set_title('Distribución de Nacimientos por Año', fontsize=14, pad=15)
axes.set_xlabel('Año', fontsize=12)
axes.set_ylabel('Cantidad de Nacimientos', fontsize=12)

```

```python
fig, axes = plt.subplots(1, figsize=(10, 4))
sns.histplot(data=df, x='grupo_etario_madre', weights='cantidad', bins=20, ax=axes, color="blue")
# Cambiamos el formato del eje Y para que no use notación científica molesta
axes.ticklabel_format(style='plain', axis='y')
axes.set_title('Distribución de Nacimientos por Grupo Etario de la Madre', fontsize=14, pad=15)
axes.set_xlabel('Grupo Etario de la Madre', fontsize=12)
axes.set_ylabel('Cantidad de Nacimientos', fontsize=12)
```

```python
df.plot(kind="line", x="anio", y="cantidad")

plt.show()
```
